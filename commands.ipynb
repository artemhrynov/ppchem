
# Bash, Git, and Conda Quick Notebook

A compact reference for common terminal and Git commands, organized for quick revision.



## 1. Moving around in Bash
Use these commands to navigate through folders.


```bash
# show current directory
pwd

# list files in current directory
ls

# move one directory up
cd ..

# move into a folder
cd myfolder
```


## 2. Copying files and folders
These commands are useful when working with files in Bash or WSL.


```bash
# copy one file into a folder
cp file1.txt myfolder/

# copy and rename at the same time
cp file1.txt myfolder/newfilename.txt

# copy all files from a folder
cp source_folder/* target_folder/

# copy a whole directory recursively
cp -r my_folder backup_folder

# show file content
cat filename.txt
```


## 3. Useful WSL tip
To open the current Linux folder directly in Windows File Explorer:


```bash
explorer.exe .
```


## 4. Git basics
A typical Git workflow is:

1. Check the status.
2. Stage files.
3. Commit.
4. Push.


```bash
# see modified, staged, and untracked files
git status

# stage one file
git add file_name

# stage all changed files
git add .

# create a commit
git commit -m "Your message"

# push committed changes to GitHub
git push
```


## 5. Working with branches
Branches let you work on changes separately from the main branch.


```bash
# show local branches
git branch

# switch to an existing branch
git switch branch_name

# create a new branch and switch to it
git switch -c new_branch_name

# connect the current local branch to the remote one
git push -u origin new_branch_name

# delete a local branch
git branch -d branch_name
```


## 6. Getting updates from GitHub
Use these when the remote repository changed.


```bash
# download and merge remote changes
git pull

# download remote changes without merging
git fetch
```


## 7. Undoing changes
Be careful with destructive commands.


```bash
# undo a specific commit safely by creating a new opposite commit
git revert COMMIT_HASH

# go back by one commit and discard changes locally
git reset --hard HEAD~1

# force-push rewritten history to GitHub
# use only if you really know you want to overwrite the remote branch
git push --force
```


## 8. Creating a `.gitignore`
Use `.gitignore` to tell Git which files or folders should not be tracked.


```bash
# create the file
touch .gitignore

# add one file to ignore
echo "secret.txt" >> .gitignore

# add one folder to ignore
echo "venv/" >> .gitignore

# ignore all log files
echo "*.log" >> .gitignore
```


If a file is **already tracked**, adding it to `.gitignore` is not enough. Remove it from tracking first:


```bash
git rm --cached secret.txt
git add .gitignore
git commit -m "Add .gitignore"
```


## 9. Conda environment export
Useful for sharing or recreating an environment.


```bash
conda env export > env.yml
```


## 10. Mini cheat sheet

- `cp` = copy
- `mv` = move
- `cd` = change directory
- `pwd` = print current directory
- `ls` = list files
- `git status` = see what changed
- `git add` = stage changes
- `git commit` = save a snapshot locally
- `git push` = send commits to GitHub
